# Palindromi
Leggi una stringa all'interno di un loop.

Se la stringa è uguale a "stop" esci dal loop.

Verifica se la stringa è un palindromo. -> Stampa "La parola è un palindromo"

Altrimenti "La parola non è un palindromo"

```python
# Leggi una stringa in input
parola = (input("Inserisci una parola: ")).lower() # trasformo la parola in minuscolo
print("La parola inserita è:", parola)
```

```python
while parola != "stop":
    # Controllo se la parola è un palindromo
    if parola == parola[::-1]:
        print(f'"{parola}" è un palindromo')
    else:
        print(f'"{parola}" non è un palindromo')

    parola = (input("Inserisci una parola: ")).lower() # trasformo la parola in minuscolo

```
