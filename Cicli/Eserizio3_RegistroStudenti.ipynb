# Registro

```python
# Chiedo quanti studenti si vogliono inserire
n = int(input("Quanti studenti vuoi inserire? "))
i = 0
studenti = {}
print(n)
```

```python
while i < n :
    dati = input("Inserisci studente: ")
    dati = dati.split()

    nome = dati[0]
    voti = list(map(float, dati[1:]))

    studenti[nome] =  voti

    i += 1

print(studenti)
```

```python
nome = input("Di chi vuoi fare la media dei voti? ")
voti = studenti[nome]
media = sum(voti) / len(voti)
print(f"{media:.2f}")
```
