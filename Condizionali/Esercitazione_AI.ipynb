In questa esercitazione svilupperai un programma che implementa diversi controlli tramite istruzioni condizionali if, if-else ed elif.

Il programma deve fare quanto segue:
1. Chiedere all'utente di inserire un'email e una password.
2. Verificare se l'email corrisponde a quella salvata (esempio: "studente@email.com"). Se non corrisponde, stampare un messaggio di accesso negato e terminare il programma.
3. Se l'email è corretta, verificare la password (esempio: "Python123"). Se la password è sbagliata, stampare un messaggio specifico di password errata.
4. Se email e password sono corrette, stampare "Accesso consentito".

Successivamente, il programma deve:</br>
5. Chiedere all'utente di inserire un numero intero.</br>
6. Verificare se il numero è:
   - pari o dispari,
   - positivo, negativo o zero.

7. Stampare un messaggio appropriato in base al valore del numero inserito, ad esempio:
   - "Il numero 4 è pari e positivo"
   - "Il numero -3 è dispari e negativo"
   - "Il numero 0 è pari e zero"

Per risolvere questo esercizio dovrai utilizzare le istruzioni condizionali, gli operatori di confronto, l'operatore modulo (%) per verificare la parità, e concatenare le condizioni con gli operatori logici AND/OR. Ricorda di usare l'indentazione corretta per definire i blocchi di codice condizionale.

Non è necessario gestire casi di input errati o eccezioni, ma ti consigliamo di provare il programma con diversi input per assicurarti che ogni condizione venga valutata correttamente.
Esempi Input/Output
input:
studente@email.com
Python123
4
output:
Accesso consentito
Il numero 4 è pari e positivo

input:
studente@email.com
passwordSbagliata
5
output:
Password errata

input:
user@altro.it
qualunque
0
output:
Accesso negato

```python
# Inizializzo le credenziali corrette come da traccia
user_email = "studente@email.com"
user_password = "Python123"
```

```python
# Chiedo i dati di autenticazione all'utente
input_email = input("Inserisci la tua mail per autenticarti: ")
input_password = input("Inserisci la tua password per autenticarti: ")
```

```python

# Controllo email
if user_email == input_email:
    # Email corretta, controllo la password
    if user_password == input_password:
        print("Accesso consentito")
        # Nota: qui devo richiedere l'input del numero perché accesso riuscito
        num = int(input("Inserisci un numero: "))
        # Analizzo il numero inserito
        if num == 0:
            print(f"Il numero 0 è pari e zero")
        elif num % 2 == 0:
            # Numero pari e non zero: controllo segno
            if num > 0:
                print(f"Il numero {num} è pari e positivo")
            else:  # num < 0
                print(f"Il numero {num} è pari e negativo")
        else:
            # Numero dispari: controllo segno
            if num > 0:
                print(f"Il numero {num} è dispari e positivo")
            else:  # num < 0
                print(f"Il numero {num} è dispari e negativo")
    else:
        # Password errata - non richiedo il numero
        print("Password errata")
else:
    # Email errata
    print("Accesso negato")
```
