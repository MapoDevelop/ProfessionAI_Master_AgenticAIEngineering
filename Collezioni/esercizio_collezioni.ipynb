# Collezioni di studenti

```python
## Ti viene fornita una lista con nomi di studenti, alcuni ripetuti, e le rispettive età in un'altra lista corrispondente (ordini paralleli)
names = ['Luca', 'Anna', 'Luca', 'Marco', 'Anna']
ages = [20, 21, 22, 23, 21]
```

```python
## Creo un dizionario che associa ciascun nome alla lista delle età corrispondenti (es. alunni con più iscrizioni o più date di nascita errate)
# Quando arrivoa un nome duplicato, il dizionario sovrascrive il dato.

student_dict = {}

for name, age in zip(names, ages):
  if name in student_dict:
    student_dict[name].append(age)
  else:
    student_dict[name] = [age]
print(f"Dizionario studenti = {student_dict}")

##Trasformare la lista dei nomi in un set, per eliminare i duplicati
# trasformo il dizionario in set per prendere i nomi
student_set = set(student_dict.keys())
print(f"Set di studenti = {student_set}")

## Creare una tupla ordinata con i nomi unici, in ordine alfabetico
# trasformo il set in tupla (ma erano già stati tolti i duplicati)
student_tupla = tuple(sorted(student_set))
print(f"Tupla di studenti = {student_tupla}")
```

```python
## Data la lista iniziale di età, dovrai creare:
## Una nuova lista dove ogni età è incrementata di 1 anno (simulare lo scorrere del tempo)
ages_new = [age + 1 for age in ages] # Incremento l'età senza creare una copia
print(f"Nuova età = {ages_new}")
```

```python
# Una copia della lista originale (notare la differenza tra assegnazione e copia con il metodo copy())
# creo la copia
ages_copy = ages.copy()
print(f"Età copiate = {ages_copy}") # Età create con copy
```

```python
# Infine, gestirai una lista annidata (lista di liste) che rappresenta le votazioni settimanali di più studenti:
#ogni elemento esterno è una lista di voti per uno studente nella settimana corrente.

weekly_grades = [[7,8,9], [6,7,7], [8,8,6]]

# Scrivi il codice per accedere e stampare il voto del terzo studente nel secondo giorno (indice 1)
initial_vote = weekly_grades[2][1] # filtro il voto studente 3 secondo giorno
print(f"Voto iniziale = {initial_vote}")

# Modifica quel voto incrementandolo di uno
modified_vote = initial_vote + 1
print(f"Voto modificato = {modified_vote}")
```
