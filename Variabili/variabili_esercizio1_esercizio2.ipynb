# **Operazioni Aritmetiche**
Inserisco due numeri interi come input e stampo:

1.   La somma dei due numeri
2.   La differenza tra i due numeri (sottrazione)
1.   Il prodotto tra due numeri





```python
# Chiedo due numeri in input
numero1 = int(input("Digita il primo numero: ")) # primo numero
numero2 = int(input("Digita il secondo numero: ")) # secondo numero

# Stampo i due numeri
print(f" Hai inserito {numero1} e {numero2}")
```

```python
# Somma i numeri
somma = numero1 + numero2

# Stampo il risultato
print(f"La somma di {numero1} + {numero2} è: {somma}")
```

```python
# Differenza tra i numeri
sottrazione = numero1 - numero2

# Stampo il risultato
print(f"La differenza tra {numero1} e {numero2} è: {sottrazione}")
```

```python
# Prodotto dei numeri
moltiplicazione = numero1 * numero2

# Stampo il risultato
print(f"il prodotto di {numero1} e {numero2} è: {moltiplicazione}")
```
