# **Potenze e Modulo**
Leggi un float a e un intero b come input e stampa:


1.   a elevato b
2.   Il modulo di a per b

All'interno del print inseriscianche l'operazioni ch stai eseguendo

esempio
3.0**2=9.0
3.0%2=1



```python
# Inserisco i dati
a = float(input("Inserisci un numero flottante: "))
b = int(input("Inserisci un numero intero: "))
```

```python
# stampo la potenza
potenza = a**b
print(f"{a:.1f}**{b}  = {potenza:.1f}")
```

```python
from IPython.core.display import b2a_base64
# stampo il modulo
modulo = a%b
print(f"{a:.1f}%{b}  = {modulo:.1f}")
```
