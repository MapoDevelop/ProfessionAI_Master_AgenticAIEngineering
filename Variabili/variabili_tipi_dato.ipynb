```python
#Recupero dei dati in input
nome = input("Come ti chiami? ") # inserimento del nome
eta = int(input("Quanti anni hai? ")) # inserimento dell'età
altezza = float(input("Quanto sei alto? ")) # inserimento dell'altezza
peso = float(input("Quanto pesi? ")) # inserimento del peso

# Calcolo dell'anno di nascita
anno_nascita = 2024 - eta # Non serve il casting perché restituisce già l'età

#Calcolo dell'BMI
bmi = float(peso/(altezza**2))

# Stampa dei dati in output
print(f"Piacere {nome}, leggo che hai {eta} e sei nato nel {anno_nascita}.")
print(f"Sei alto {altezza:.2f} e pesi {peso:.2f}, il tuo BMI è {bmi:.2f}")
```
