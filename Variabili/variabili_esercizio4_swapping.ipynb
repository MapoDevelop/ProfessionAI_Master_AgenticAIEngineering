# ***Swapping***
Lo swapping consiste nell'invertire il valore di due variabili.



1.   Prendi in imput due valori e assegnali ad A e B
2.   Stampa le variabili su schermo


1.   Assegna il valore di A a B e il valore di B ad A
2.   Stampa le variabili su schermo





```python
import numpy as np

# Inizializzo le variabili
a = input("Inserisci un valore a A: ")
b = input("Inserisci un valore a B: ")

print(f"A = {a} e B = {b}")
```

```python
# inverto i due valori
a , b = b , a

print(f"A = {a} e B = {b}")
```
